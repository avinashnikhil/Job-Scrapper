```python
import requests
from bs4 import BeautifulSoup
import csv
from google.colab import files

def scrape_fake_jobs(url="https://realpython.github.io/fake-jobs/", output_filename="fake_jobs.csv"):
    try:
        # Fetch the webpage content
        response = requests.get(url, timeout=10)
        response.raise_for_status()
    except requests.exceptions.RequestException as e:
        print(f"Error fetching the webpage: {e}")
        return

    # Parse the HTML content
    soup = BeautifulSoup(response.content, "html.parser")
    job_cards = soup.find_all("div", class_="card-content")

    if not job_cards:
        print("No jobs found. The webpage structure might have changed.")
        return

    jobs_data = []

    for card in job_cards:
        # Extract fields with safe fallbacks for missing data
        title_elem = card.find("h2", class_="title")
        title = title_elem.text.strip() if title_elem else "N/A"

        company_elem = card.find("h3", class_="company")
        company = company_elem.text.strip() if company_elem else "N/A"

        location_elem = card.find("p", class_="location")
        location = location_elem.text.strip() if location_elem else "N/A"

        # Extract Job Detail Page URL (specifically looking for the 'Apply' link)
        job_url = "N/A"
        for link in card.find_all("a"):
            if link.text.strip().lower() == "apply":
                job_url = link.get("href", "N/A")
                break

        jobs_data.append({
            "Job Title": title,
            "Company": company,
            "Location": location,
            "Job Detail URL": job_url
        })

    # Save to CSV in Colab's virtual environment
    with open(output_filename, mode="w", newline="", encoding="utf-8") as file:
        writer = csv.DictWriter(file, fieldnames=["Job Title", "Company", "Location", "Job Detail URL"])
        writer.writeheader()
        writer.writerows(jobs_data)

    print(f"Successfully scraped {len(jobs_data)} jobs!")

    # Automatically trigger the download to your local machine
    print("Downloading CSV file...")
    files.download(output_filename)

# Run the scraper
scrape_fake_jobs()
```
